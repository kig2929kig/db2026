
# 6. SQL 실행을 도와주는 실습 함수 준비

이제부터는 SQL을 여러 줄로 작성해야 합니다.

아래 `run_sql()` 함수는 학생이 작성한 SQL 문자열을 MySQL에 전달해 실행해 줍니다.

> 이 셀은 **수정하지 말고 한 번 실행**하세요.

또한 SQL 안에 `TODO`가 남아 있으면 실행하지 않고 알려주도록 만들었습니다.


````python
import subprocess
import textwrap

def run_sql(sql):
    sql = textwrap.dedent(sql).strip()

    if not sql:
        print("⚠️ SQL이 비어 있습니다.")
        return

    if "TODO" in sql:
        print("⚠️ SQL 안에 TODO가 남아 있습니다. 먼저 완성하세요.")
        return

    result = subprocess.run(
        ["sudo", "mysql"],
        input=sql,
        text=True,
        capture_output=True
    )

    if result.stdout:
        print(result.stdout)

    if result.stderr:
        print("❌ MySQL 오류:")
        print(result.stderr)
    elif result.returncode == 0:
        print("✅ SQL 실행 완료")
````

데이터베이스가 '서류함'이라면 테이블은 실제 데이터를 저장하는 **표**에 가깝습니다.

이번에 만들 `students` 테이블의 구조는 다음과 같습니다.

| 컬럼 | 자료형 | 역할 |
|---|---|---|
| `id` | INT | 학생을 구분하는 번호 |
| `name` | VARCHAR(50) | 이름 |
| `grade` | INT | 학년 |
| `class_name` | VARCHAR(50) | 반 또는 학과 |
| `created_at` | TIMESTAMP | 데이터 생성 시각 |

## 꼭 알아야 할 제약조건

### PRIMARY KEY
각 레코드를 유일하게 구분하는 **기본키**입니다.  
같은 기본키 값이 중복될 수 없습니다.

### AUTO_INCREMENT
새 레코드가 들어올 때 `id` 값을 1, 2, 3...처럼 자동으로 증가시킵니다.

### NOT NULL
해당 컬럼을 비워 둘 수 없다는 뜻입니다.

### DEFAULT CURRENT_TIMESTAMP
값을 직접 입력하지 않아도 현재 날짜와 시간이 자동으로 저장됩니다.

> 중복으로 INSERT 셀을 여러 번 실행했다면 레코드 수가 5개보다 많아질 수 있습니다.  
> 그런 경우 아래의 **초기화 방법**을 참고하세요.


````python
import subprocess

def mysql_value(query):
    result = subprocess.run(
        ["sudo", "mysql", "-N", "-B", "-e", query],
        text=True,
        capture_output=True
    )
    if result.returncode != 0:
        return None
    return result.stdout.strip()

db_exists = mysql_value(
    "SELECT COUNT(*) FROM information_schema.SCHEMATA WHERE SCHEMA_NAME='school';"
)
table_exists = mysql_value(
    "SELECT COUNT(*) FROM information_schema.TABLES "
    "WHERE TABLE_SCHEMA='school' AND TABLE_NAME='students';"
)
column_count = mysql_value(
    "SELECT COUNT(*) FROM information_schema.COLUMNS "
    "WHERE TABLE_SCHEMA='school' AND TABLE_NAME='students';"
)

print("===== 자동 점검 =====")
print("school 데이터베이스 :", "✅" if db_exists == "1" else "❌")
print("students 테이블     :", "✅" if table_exists == "1" else "❌")
print("컬럼 5개            :", "✅" if column_count == "5" else f"❌ ({column_count})")

if table_exists == "1":
    record_count = mysql_value("SELECT COUNT(*) FROM school.students;")
    grade3_count = mysql_value("SELECT COUNT(*) FROM school.students WHERE grade=3;")
    print("학생 레코드 5개     :", "✅" if record_count == "5" else f"⚠️ ({record_count}개)")
    print("3학년 학생 2명      :", "✅" if grade3_count == "2" else f"⚠️ ({grade3_count}명)")
else:
    print("학생 레코드 5개     : 확인 불가")
    print("3학년 학생 2명      : 확인 불가")
````
