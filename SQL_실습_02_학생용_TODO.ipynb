
# 2. SQL 실행 함수 준비

여러 줄의 SQL을 편하게 실행할 수 있도록 도우미 함수를 준비합니다.

아래 셀은 수정하지 마세요.


````python
import subprocess
import textwrap

def run_sql(sql):
    sql = textwrap.dedent(sql).strip()

    if not sql:
        print("⚠️ SQL이 비어 있습니다.")
        return

    if "TODO" in sql:
        print("⚠️ SQL 안에 TODO가 남아 있습니다. 먼저 완성하세요.")
        return

    result = subprocess.run(
        ["sudo", "mysql"],
        input=sql,
        text=True,
        capture_output=True
    )

    if result.stdout:
        print(result.stdout)

    if result.stderr:
        print("❌ MySQL 오류:")
        print(result.stderr)
    elif result.returncode == 0:
        print("✅ SQL 실행 완료")
````

# 5. CREATE DATABASE 기본 문법

가장 기본적인 데이터베이스 생성 문법은 다음과 같습니다.

```sql
CREATE DATABASE 데이터베이스이름;
```

예:

```sql
CREATE DATABASE school;
```

하지만 한글과 다양한 문자를 안정적으로 저장하기 위해 문자 설정도 함께 지정할 수 있습니다.

```sql
CREATE DATABASE school
DEFAULT CHARACTER SET utf8mb4
DEFAULT COLLATE utf8mb4_unicode_ci;
```

---

## CHARACTER SET

문자를 **어떤 방식으로 저장할 것인지** 정합니다.

이번 실습에서 사용하는 `utf8mb4`는 한글, 영어, 숫자뿐 아니라 다양한 유니코드 문자와 이모지까지 저장할 수 있습니다.

## COLLATE

문자열을 **비교하고 정렬할 때 어떤 규칙을 적용할지** 정합니다.

쉽게 기억하면:

- `CHARACTER SET` = 글자를 저장하는 규칙
- `COLLATE` = 글자를 비교·정렬하는 규칙


# 10. 테이블이란?

테이블은 실제 데이터를 저장하는 **행(row)과 열(column) 구조의 표**입니다.

예를 들어 `students` 테이블은 다음처럼 생각할 수 있습니다.

| id | name | grade | class_name |
|---:|---|---:|---|
| 1 | 김부장 | 3 | 컴퓨터과 |
| 2 | 혼이 | 1 | 영혼반 |

- **행(row)** : 학생 한 명의 전체 정보
- **열(column)** : 이름, 학년, 반처럼 각 항목
- **테이블(table)** : 여러 행과 열로 구성된 저장 공간

---

## 꼭 기억하기

> 학생 한 명의 정보가 한 **행**이 되고,  
> 이름·학년·반 같은 항목이 **열**이 됩니다.

# 12. id 컬럼 이해하기

```sql
id INT AUTO_INCREMENT PRIMARY KEY
```

### INT
정수형 숫자를 저장합니다.

### PRIMARY KEY
각 행을 서로 구분하는 **고유한 값**입니다.

학생 이름이 같아도 `id` 값은 서로 달라야 합니다.

### AUTO_INCREMENT
새 레코드가 추가될 때 번호를 자동으로 증가시킵니다.

예:

```text
1
2
3
4
...
```

따라서 학생 데이터를 입력할 때 `id`를 직접 입력하지 않아도 됩니다.

SQL에서는 이름을 알아보기 쉽게 정하는 습관이 중요합니다.

### 추천하는 방식

- 데이터베이스명: 소문자 사용
- 테이블명: 복수형 또는 의미 있는 이름

- 공백 사용하지 않기
- 한글 이름은 가능하면 피하기
- SQL 예약어는 피하기

좋은 예:

```text
school
students
class_name
created_at
```

피하는 것이 좋은 예:

```text
학생 정보
student name
1grade
select
order
```

`SELECT`, `ORDER`, `TABLE` 같은 단어는 SQL에서 이미 특별한 의미를 가지므로 이름으로 사용하지 않는 것이 좋습니다.


# 23. 데이터베이스를 잘못 만들었을 때

데이터베이스 자체를 삭제하는 명령은 다음과 같습니다.

```sql
DROP DATABASE school;
```

하지만 이 명령은 `school` 안의 모든 테이블과 데이터까지 삭제합니다.

```sql

```

> **주의:**  
> `DROP DATABASE`는 큰 서류함 하나를 통째로 버리는 것과 같습니다.  
> 안의 테이블과 데이터도 함께 사라집니다.

이 노트북에서는 학습을 위해 설명만 하고 바로 삭제하지 않습니다.


# 25. 자동 점검

아래 셀은 학생의 SQL 정답을 보여 주지 않고,  
현재 데이터베이스와 테이블 구조가 수업 목표에 맞는지만 확인합니다.

정상적으로 만들었다면 대부분 ✅가 표시됩니다.


````python
import subprocess

def mysql_value(query):
    result = subprocess.run(
        ["sudo", "mysql", "-N", "-B", "-e", query],
        text=True,
        capture_output=True
    )
    if result.returncode != 0:
        return None
    return result.stdout.strip()

checks = {}

checks["school DB"] = mysql_value(
    "SELECT COUNT(*) FROM information_schema.SCHEMATA WHERE SCHEMA_NAME='school';"
)

checks["students table"] = mysql_value(
    "SELECT COUNT(*) FROM information_schema.TABLES "
    "WHERE TABLE_SCHEMA='school' AND TABLE_NAME='students';"
)

checks["column count"] = mysql_value(
    "SELECT COUNT(*) FROM information_schema.COLUMNS "
    "WHERE TABLE_SCHEMA='school' AND TABLE_NAME='students';"
)

print("===== 자동 점검 =====")
print("school 데이터베이스 :", "✅" if checks["school DB"] == "1" else "❌")
print("students 테이블     :", "✅" if checks["students table"] == "1" else "❌")
print("컬럼 5개            :", "✅" if checks["column count"] == "5" else f"❌ ({checks['column count']})")

if checks["students table"] == "1":
    pk = mysql_value(
        "SELECT COUNT(*) FROM information_schema.COLUMNS "
        "WHERE TABLE_SCHEMA='school' AND TABLE_NAME='students' "
        "AND COLUMN_NAME='id' AND COLUMN_KEY='PRI';"
    )
    ai = mysql_value(
        "SELECT COUNT(*) FROM information_schema.COLUMNS "
        "WHERE TABLE_SCHEMA='school' AND TABLE_NAME='students' "
        "AND COLUMN_NAME='id' AND EXTRA LIKE '%auto_increment%';"
    )
    notnull = mysql_value(
        "SELECT COUNT(*) FROM information_schema.COLUMNS "
        "WHERE TABLE_SCHEMA='school' AND TABLE_NAME='students' "
        "AND COLUMN_NAME='name' AND IS_NULLABLE='NO';"
    )
    print("id PRIMARY KEY      :", "✅" if pk == "1" else "❌")
    print("id AUTO_INCREMENT   :", "✅" if ai == "1" else "❌")
    print("name NOT NULL       :", "✅" if notnull == "1" else "❌")
````


# 26. 오늘 배운 내용 정리

이번 실습의 핵심 명령어:

```sql
CREATE DATABASE
USE
CREATE TABLE
SHOW TABLES
DESC
DROP TABLE IF EXISTS
DROP DATABASE IF EXISTS
```

핵심 개념:

```text
데이터베이스 = 여러 테이블을 담는 큰 보관함
테이블       = 행과 열로 구성된 데이터 저장 공간
행(row)      = 하나의 실제 데이터
열(column)   = 데이터의 항목
```

제약조건과 옵션:

```text
PRIMARY KEY        : 각 행을 구분하는 고유한 값
AUTO_INCREMENT     : 숫자를 자동으로 증가
NOT NULL           : 빈 값 허용 안 함
DEFAULT            : 값을 생략했을 때 사용할 기본값
CURRENT_TIMESTAMP  : 현재 날짜와 시간
```
